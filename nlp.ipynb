```python
import re
```

```python
text="paragraph the cow  cow is a gentle domestic animal found worldwide in various colors, shapes, and sizes. It has a large body with four legs, two ears, two eyes, a big nose, a long tail, and often two sharp horns. As a herbivore, the cow primarily eats grass, hay, grains, and vegetables to stay healthy"
```

```python
token= word_tokenize(text)
token
stop_rem=[serial.lower() for serial in token if serial not in stop_word ]
stop_rem
lammati=[lemma.lemmatize(word) for word in stop_rem]
lmt=" ".join(lammati)
clean_text = re.sub(r"[^\w\s]", "", lmt).strip()
clean_numbrt = re.sub(r"\d+", "", clean_text)
clean_numbrt=list([clean_numbrt])
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv=CountVectorizer()
```

```python
model=cv.fit_transform(clean_numbrt)

model
```

```python
model.toarray()
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

def create_cloud(string, title=None):
    cloud = WordCloud(
        width=800,
        height=400,
        background_color='white'
    ).generate(string)

    plt.figure(figsize=(10, 5))
    plt.imshow(cloud, interpolation='bilinear')
    plt.axis("off")

    if title:
        plt.title(title)

    plt.show()

# function call
create_cloud(" ".join(pos_sen), title="My Word Cloud")
```
